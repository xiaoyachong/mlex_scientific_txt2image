```python
import json
import codecs
import ast
import shutil
with open("xc_empty_5k_2_1000fake_1000realexp_3rd.txt", "r", encoding='utf-8') as f:  
    data = f.read()  
    
json_string=data[1:-2]
json_data = ast.literal_eval(json_string)
print(len(json_data))
print(json_data[0])

fake_list=[]
real_list=[]
for i in range(len(json_data)):
    filename = json_data[i]['uri'].split("/")[-1]
    
    label = json_data[i]['tags'][-1]['name']
    if label =="Fake":
        fake_list.append(filename)
    elif label =="Real":
        real_list.append(filename)

print(len(fake_list),len(real_list))

for filename in fake_list:
    src_fake_path = "/data/mlex_repo/mlex_labelmaker_v2/dash_labelmaker_demo/data/empty_5k_2/" + filename
    dst_fake_path = "/lovelace/xiaoya/GenAI_dataset/xc_empty_5k_2_1000fake_1000realexp_3rd/Fake/"+ filename
    shutil.copyfile(src_fake_path, dst_fake_path)


for filename in real_list:
    src_real_path = "/data/mlex_repo/mlex_labelmaker_v2/dash_labelmaker_demo/data/empty_5k_2/" + filename
    dst_real_path = "/lovelace/xiaoya/GenAI_dataset/xc_empty_5k_2_1000fake_1000realexp_3rd/Real/"+ filename
    shutil.copyfile(src_real_path, dst_real_path)
```
